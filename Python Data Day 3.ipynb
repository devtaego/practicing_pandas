```python
import requests
from bs4 import BeautifulSoup


def get_newsDataAI(keyword, page=1):

    url = "https://search.daum.net/search"

    params = {
        "w": "tot",
        "DA": "YZR",
        "t__nil_searchbox": "btn",
        "q": keyword,
        "p": page
    }

    headers = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/140.0.0.0 Safari/537.36"
        )
    }

    response = requests.get(
        url,
        params=params,
        headers=headers
    )

    print("응답 코드 :", response.status_code)

    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    # 뉴스 기사 찾기
    news_items = soup.select("div.item-bundle-mid")

    print("찾은 뉴스 개수 :", len(news_items))

    news_list = []

    for item in news_items:

        title_tag = item.select_one("div.item-title a")
        summary_tag = item.select_one(
            "div.item-contents p.conts-desc a"
        )

        if title_tag is None:
            continue

        title = title_tag.get_text(" ", strip=True)
        link = title_tag.get("href")

        if summary_tag:
            summary = summary_tag.get_text(" ", strip=True)
        else:
            summary = ""

        news_list.append({
            "title": title,
            "link": link,
            "summary": summary
        })

    return news_list
```

```python
import requests
from bs4 import BeautifulSoup


def get_bookData():

    url = "https://www.yes24.com/product/category/bestseller"

    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    book_list = []

    # 1~100위
    for page in range(1, 6):

        params = {
            "categoryNumber": "001001025",
            "pageNumber": page
        }

        response = requests.get(
            url,
            params=params,
            headers=headers
        )

        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        # 책 하나씩
        books = soup.select("div.itemUnit")

        for book in books:

            # 책 제목
            title_tag = book.select_one(
                "a.gd_name"
            )

            if title_tag is None:
                continue

            title = title_tag.get_text(
                " ", strip=True
            )

            # 저자
            author_tag = book.select_one(
                "span.info_auth"
            )

            author = (
                author_tag.get_text(" ", strip=True)
                if author_tag
                else ""
            )

            # 출판사
            publisher_tag = book.select_one(
                "span.info_pub"
            )

            publisher = (
                publisher_tag.get_text(" ", strip=True)
                if publisher_tag
                else ""
            )

            # 출판일
            date_tag = book.select_one(
                "span.info_date"
            )

            publish_date = (
                date_tag.get_text(" ", strip=True)
                if date_tag
                else ""
            )

            # 가격
            price_tag = book.select_one(
                "strong.txt_num"
            )

            price = (
                price_tag.get_text(" ", strip=True)
                if price_tag
                else ""
            )

            # 평점
            rating_tag = book.select_one(
                "span.rating_grade em"
            )

            rating = (
                rating_tag.get_text(" ", strip=True)
                if rating_tag
                else ""
            )

            book_data = {
                "title": title,
                "author": author,
                "publisher": publisher,
                "publish_date": publish_date,
                "price": price,
                "rating": rating
            }

            book_list.append(book_data)

    return book_list[:100]
```

```python
import requests
from bs4 import BeautifulSoup
import json


def get_jobData(page=1):

    url = "https://www.saramin.co.kr/zf_user/jobs/list/job-category"

    params = {
        "cat_kewd": "87",
        "recruitPage": page
    }

    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    response = requests.get(
        url,
        params=params,
        headers=headers
    )

    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    job_data = []

    # 공고 하나 = list_item 하나
    jobs = soup.select(".list_item")

    print("찾은 공고:", len(jobs))

    for job in jobs:

        # -------------------------
        # 회사명
        # -------------------------

        company_tag = job.select_one(
            ".company_nm .str_tit"
        )

        company = ""

        if company_tag:
            company = company_tag.get_text(
                " ",
                strip=True
            )

        # -------------------------
        # 공고 제목 + 링크
        # -------------------------

        title_tag = job.select_one(
            ".job_tit a"
        )

        title = ""
        link = ""

        if title_tag:

            title = title_tag.get_text(
                " ",
                strip=True
            )

            link = title_tag.get("href", "")

            if link.startswith("/"):
                link = "https://www.saramin.co.kr" + link

        # -------------------------
        # 지역
        # -------------------------

        location_tag = job.select_one(
            ".work_place"
        )

        location = ""

        if location_tag:
            location = location_tag.get_text(
                " ",
                strip=True
            )

        # -------------------------
        # 경력
        # -------------------------

        career_tag = job.select_one(
            ".career"
        )

        career = ""

        if career_tag:
            career = career_tag.get_text(
                " ",
                strip=True
            )

        # -------------------------
        # 학력
        # -------------------------

        education_tag = job.select_one(
            ".education"
        )

        education = ""

        if education_tag:
            education = education_tag.get_text(
                " ",
                strip=True
            )

        # -------------------------
        # 마감일
        # -------------------------

        deadline_tag = job.select_one(
            ".support_detail .date"
        )

        deadline = ""

        if deadline_tag:
            deadline = deadline_tag.get_text(
                " ",
                strip=True
            )

        # -------------------------
        # 데이터 저장
        # -------------------------

        data = {
            "company": company,
            "title": title,
            "location": location,
            "career": career,
            "education": education,
            "deadline": deadline,
            "link": link
        }

        job_data.append(data)

    return job_data


# =====================================
# 1페이지 크롤링
# =====================================

jobs = get_jobData(page=1)

print("공고 개수:", len(jobs))


# 처음 5개 출력
for job in jobs[:5]:
    print(job)


# =====================================
# JSON 저장
# =====================================

with open(
    "saramin_jobs.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        jobs,
        f,
        ensure_ascii=False,
        indent=4
    )

print("JSON 저장 완료")
```
