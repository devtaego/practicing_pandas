```python
## Pandas
- 데이터 분석에 적합한 새로운 자료구조 2개를 제공
- Series, DataFrame
```

```python
import pandas as pd
```

```python
### Series
- 1차원 자료구조
- 구성요소 2개 : index, values
- 순서있음 => 슬라이싱, ...
```

```python
# A, B, C, D, E
math = [90,89,91,92,90]

math_sr = pd.Series(math, index=['A','B','C','D','E'])
```

```python
# 데이터 접근 => 인덱스
math_sr['A']
```

```python
# 순서로도 접근가능
math_sr.iloc[0]
```

```python
# 슬라이싱 => 범위로 가져오기
math_sr[:2]
```

```python
math_sr[-2:]
```

```python
math_sr.sort_values()
```

```python
math_sr.sort_values(ascending=False)
```

```python
# 필터링
# 조건에 맞는 데이터만 가져오기
math_sr[ math_sr > 90 ]
```

```python
eng = [91,92,90,91,89]

eng_sr = pd.Series(eng, index=['A','B','C','D','E'])
```

```python
math_sr + eng_sr
math_sr - eng_sr
math_sr * eng_sr
math_sr / eng_sr
math_sr // eng_sr
math_sr % eng_sr
```

```python
# 시리즈와 값 하나 연산
# 데이터 타입에 맞춰서 (인덱스의) 값 하나를 다 연산시켜 줌
math_sr + 100
```

```python
(math_sr + eng_sr ) / 2
```

```python
### DataFrame
- 2차원 자료구조
- 구성요소 3개 : column, index, values
- 순서는 기본적으로 행방향
```

```python
# A, B, C, D, E
math = [90,89,91,92,90]
eng = [91,92,90,91,89]
kor = [90,91,92,90,89]

grade_df = pd.DataFrame({'math':math, 'kor':kor, 'eng':eng}, index=['A','B','C','D','E'])
```

```python
# 데이터 접근하기
# 기본적으로 컬럼
grade_df['math']

# 결과를 보면 series형태로 나옴
```

```python
# 한번에 컬럼을 여러개를 선택하려면 => 리스트로 묶어서 하나로 입력 ★

grade_df[['math', 'eng']]
```

```python
# []
1. 리스트 생성
2. 자료구조에서 데이터 꺼낼 때 ★★ => 인덱싱

딕셔너리[키]
리스트[순서]
```

```python
# 인덱스를 기준으로 선택
# .loc[인덱스]

grade_df.loc['A']
```

```python
# df에서 바로 한 값에 접근하기
# .loc[인덱스, 컬럼]

grade_df.loc[ 'A', 'math']
```

```python
grade_df['math']['A'] 
# grade_df['math']['A'] = 100
```

```python
grade_df.loc['A', 'math'] = 100
```

```python
# grade_df['math']['A'] 
grade_df.loc['A', 'math']
```

```python
# 행과 열 순서
# .iloc[행 순서, 열 순서]

grade_df.iloc[1,0]
```
