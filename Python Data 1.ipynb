```python
### DataFrame
- 2차원 자료구조
- 구성요소 3개 : column, index, values
- 순서는 기본적으로 행방향
```

```python
# 필터링
# 1. 조건을 줄 때는 어떤 컬럼을 기준으로 조건을 주는지 명확히!

# grade_df[ grade_df > 90 ] !error! 
grade_df[ grade_df['math'] > 90 ]
```

```python
# 2. 조건 2개일 때~
# 논리연산 : and(&), or(|), not(~)

grade_df[ (grade_df['math'] >= 90) & grade_df['eng'] >= 90]
```

```python
# 3. 조건이 3개 이상일 때
# 비교연산을 먼저 하고 변수에 할당시켜서 사용

grade_df[ (grade_df['math']>=90) &  (grade_df['eng']>=90) &  (grade_df['kor']>=90) ] # 좋지 않음
```

```python
# 더 나은 방식
cond1 = grade_df['math'] >= 90
cond2 = grade_df['eng'] >= 90
cond3 = grade_df['kor'] >= 90
grade_df[ cond1 & cond2 & cond3 ]
```

```python
# 시리즈의 연산
# df에서 어떤 속성을 선택해서 연산을 시키고 새로운 속성을 계산함
# => 기존의 df에 붙여서 같이 분석해 줌

grade_df['total'] = grade_df['math'] + grade_df['eng'] + grade_df['kor']  
```

```python
grade_df['avg'] = grade_df['total'] / 3
```
